```python
from pathlib import Path
```

```python
def add_document_mount(source_path: str, target_path: str) -> None:
    """
    Adds a document mount to the docker-compose.yml file.

    Args:
        source_path (str): The source path of the documents on the server (absolute or relative path).
        target_path (str): The target path inside the container (relative to app/documents inside the container).
    """
    with open("docker-compose.yml", "r+") as f:
        lines = f.readlines()

        for i, line in enumerate(lines):
            if line.strip() == "# - DOCUMENT MOUNTS -":
                spaces = line.find("#")
                lines.insert(i + 1, " " * spaces + "- type: bind\n")
                lines.insert(i + 2, " " * spaces + f"  source: {source_path}\n")
                lines.insert(i + 3, " " * spaces + "  target: ${MOUNT_PATH}/" + f"{target_path}\n")
                break

        f.seek(0)
        f.writelines(lines)
        f.truncate()
```
